```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Algerian_forest_fires_dataset.csv')
```

```python
df.head()
```

```python
df.info
```

```python
##missing values

df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,"Region"]=0
df.loc[122:,"Region"]=1
```

```python
df.head()
```

```python
df.info()
```

```python
df["Region"]=df["Region"].astype(int)
```

```python
df.isnull().sum()
```

```python
#Remove the null values
df=df.dropna()
```

```python
df.isnull().sum()
```

```python
df=df.reset_index()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
#fix spaces in the column names
df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
df.info()
```

```python
df.info()
```

```python
#change the required columns as integer data type
df[[ 'Temperature', 'RH', 'Ws', 'Rain',
       'FFMC', 'DMC', 'DC', 'ISI', 'BUI', 'FWI']]=df[[ 'Temperature', 'RH', 'Ws', 'Rain',
       'FFMC', 'DMC', 'DC', 'ISI', 'BUI', 'FWI']].astype(float)
```

```python
df[['month', 'year', 'Temperature', 'RH', 'Ws']]=df[['month', 'year', 'Temperature', 'RH', 'Ws']].astype('Int32')
```

```python
##changing the other columns to float data type
objects=[feature for feature in df.columns if df[feature].dtypes=='O']
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
df.head()
```

```python
#save the cleaned dataset into neew csv file
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```
